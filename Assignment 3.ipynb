```python
def is_right_triangle(a, b, c):
    """
    Checks if three sides form a right-angled triangle.
    """
    sides = [a, b, c]
    sides.sort()
    return sides[0]**2 + sides[1]**2 == sides[2]**2

def main():
    print("--- Right-Angled Triangle Checker ---")
    try:
        side1 = float(input("Enter the first side: "))
        side2 = float(input("Enter the second side: "))
        side3 = float(input("Enter the third side: "))
        if side1 <= 0 or side2 <= 0 or side3 <= 0:
            print("Invalid input: Triangle sides must be greater than zero.")
        else:
            if is_right_triangle(side1, side2, side3):
                print("Result: The triangle IS a right-angled triangle.")
            else:
                print("Result: The triangle is NOT a right-angled triangle.")
                
    except ValueError:
        print("Invalid input: Please enter numerical values only.")

if __name__ == "__main__":
    main()
```
